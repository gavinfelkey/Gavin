```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def prime_star_systems(arr):
    vectorized_is_prime = np.vectorize(is_prime)
    mask = vectorized_is_prime(arr)
    row_mask = mask.any(axis = 1)
    return arr[row_mask]

print(prime_star_systems(arr))
```

```python
def expansion(arr, num_spaces):
    spacer = " " * num_spaces

    def expand_single_string(s):
        return spacer.join(s)
    vectorized_expansion = np.vectorize(expand_single_string)
    return vectorized_expansion(arr)

universe = np.array(['galaxy', 'clusters'])

print("Result for 1 space:")
print(expansion(universe, 1))

print("\nResult for 2 spaces:")
print(expansion(universe, 2))
```

```python
def secondDimmest(arr):
    sorted_stars = np.sort(arr, axis=1)
    return sorted_stars[:, -2]

np.random.seed(123)
arr = np.random.randint(500, 2000, (5, 5))

print(arr)

result = secondDimmest(arr)
print(result)
```
