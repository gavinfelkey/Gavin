```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def make_sin_wave(x, A, w):
    return A * np.sin(w * x)
x = np.linspace(0, 2 * np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
linestyles = ['-', '--', ':', '-.', '-.']
```

```python
fig, axes = plt.subplots(1, 3, figsize = (25, 5))
for idx, (A, w) in enumerate(zip(amplitude, frequency)):
    ls = linestyles[idx]
    y1 = make_sin_wave(x, A, 1)
    axes[0].plot(x, y1, linestyle = ls, label=f'Amplitude={A}, Frequency=1')
    y2 = make_sin_wave(x, 1, w)
    axes[1].plot(x, y2, linestyle = ls, label=f'Amplitude={A}, Frequency=1')
    y3 = make_sin_wave(x, A, w)
    axes[2].plot(x, y3, linestyle = ls, label=f'Amplitude={A}, Frequency=1')
axes[0].set_title('Varying Amplitude Only', fontsize=16)
axes[0].set_xlabel('x (radians)', fontsize=14)
axes[0].set_ylabel('y', fontsize=14)
axes[0].legend(fontsize=12, loc='upper right')
axes[1].set_title('Varying Frequency Only', fontsize=16)
axes[1].set_xlabel('x (radians)', fontsize=14)
axes[1].set_ylabel('y', fontsize=14)
axes[1].legend(fontsize=12, loc='upper right')
axes[2].set_title('Varying Amplitude and Frequency Together', fontsize=16)
axes[2].set_xlabel('x (radians)', fontsize=14)
axes[2].set_ylabel('y', fontsize=14)
axes[2].legend(fontsize=12, loc='upper right')
plt.tight_layout()
plt.show()
```
