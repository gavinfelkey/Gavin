```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]

min_mag = np.min(magnitude)
max_mag = np.max(magnitude)
print(f"Minimum Magnitude: {min_mag}")
print(f"Maximum Magnitude: {max_mag}")

min_temp = np.min(temperature)
max_temp = np.max(temperature)
print(f"Minimum Temperature: {min_temp}")
print(f"Maximum Temperature: {max_temp}")

brightest_idx = np.argmin(magnitude)
dimmest_idx = np.argmax(magnitude)
print(f"Brightest Star Index: {brightest_idx} (Magnitude: {magnitude[brightest_idx]})")
print(f"Dimmest Star Index: {dimmest_idx} (Magnitude: {magnitude[dimmest_idx]})")

coldest_idx = np.argmin(temperature)
hottest_idx = np.argmax(temperature)
print(f"Hottest star: {hottest_idx} (Temperature: {temperature[hottest_idx]})")
print(f"Coldest star: {coldest_idx} (Temperature: {temperature[coldest_idx]})")
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
plt.figure(figsize=(8, 6))
scatter = plt.scatter(log_temperature, log_magnitude, c=temperature, label="Stars")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.scatter(log_temperature[brightest_idx], log_magnitude[brightest_idx], 
            color='black', marker='*', s=300, label='Brightest Star')
plt.scatter(log_temperature[dimmest_idx], log_magnitude[dimmest_idx], 
            color='purple', marker='*', s=300, label='Dimmest Star')
plt.scatter(log_temperature[hottest_idx], log_magnitude[hottest_idx], 
            color='green', marker='*', s=300, label='Hottest Star')
plt.scatter(log_temperature[coldest_idx], log_magnitude[coldest_idx], 
            color='orange', marker='*', s=300, label='Coldest Star')
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.title("Hertzsprung-Russell Diagram")
cbar = plt.colorbar(scatter)
cbar.set_label('Temperature (K)')
plt.legend()
plt.show()
```
